# Part 1 Single Prediction

The weight scales each blade angle down by a half, essentially making it 
less "important" for the predicition.

```python
blade_angle = [8.5, 9.5, 9.9, 9.0]
weight = 0.5

def neural_network(input, input_weight):
    return input * input_weight

for angle in blade_angle:
    prediction = neural_network(angle, weight)
    print(f"Prediction for {angle}: {prediction}")
```

# Part 2 Multiple Inputs

Real predictions move beyond single features by using multiple inputs (specifically blade angle, balance, and breath tempo) to make a single prediction using a weighted sum (dot product). The w_sum function takes two lists of equal length, enforced by an assert statement, and multiplies each input by its matching weight before adding the products together. The NumPy implementation performs this same weighted sum using np.array and .dot(), which multiplies corresponding values and sums the results. With only three terms, the from-scratch and NumPy versions match exactly. However, on a calculation with ten thousand terms, they might differ slightly in the final decimal place because NumPy may add terms in a different order, which is why floating-point values are compared using a tolerance (such as 1e-9) rather than ==.

```python
import numpy as np
#
weights = [0.1, 0.2, 0.0]

blade_angle = [8.5, 9.5, 9.9, 9.0]
balance = [0.65, 0.80, 0.80, 0.90]
breath = [1.2, 1.3, 0.5, 1.0]


def w_sum(a, b):
    assert len(a) == len(b)

    total = 0      

    for i in range(len(a)):
        total += a[i] * b[i]

    return total

def neural_network(input, weights):
    return w_sum(input, weights)

for i in range(4):
    input = [blade_angle[i], balance[i], breath[i]]
    prediction = neural_network(input, weights)
    print(f"sensing {i}: {prediction}")

# -- numpy version --

def neural_network_numpy(input, weights):
    input = np.array(input)
    weights = np.array(weights)


    return input.dot(weights)

for i in range(4):
    input = [blade_angle[i], balance[i], breath[i]]

    scratch_prediction = neural_network(input, weights)
    numpy_prediction = neural_network_numpy(input, weights)

    print(
        f"sensing {i}: "
        f"from scratch = {scratch_prediction}, "
        f"numPy = {numpy_prediction}, "
        f"match = {abs(scratch_prediction - numpy_prediction) < 1e-9}"
    )
```

# Part 3 Multiple Outputs

Part 3 handles multiple outputs by taking a single input, such as stance balance, and multiplying it across a vector of weights to generate predictions for whether the partner opens left, strikes high, or feints. The ele_mul function accomplishes this from scratch by multiplying the scalar input by each weight in the vector. In NumPy, multiplying a scalar directly by an ndarray performs the same elementwise multiplication without an explicit Python loop. This works because NumPy uses broadcasting to automatically scale the scalar across all dimensions of the array in optimized C code.
